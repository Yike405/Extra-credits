```python
# check grades above 100 — may reflect extra credit or curved grading
grade_cols = ['autumn', 'winter', 'spring', 'summer']

for col in grade_cols:
    above = df[df[col] > 100][['student_ID', 'subject', col]]
    print(f'{col} > 100: {len(above)} rows')
    print(above.to_string())
    print()
```

## 2. Data Cleaning

```python
# standardize column names to lowercase
df.columns = df.columns.str.strip().str.lower()
print(df.columns.tolist())
```

```python
# standardize subject names to lowercase
df['subject'] = df['subject'].str.strip().str.lower()
df['subject'].unique()
```

```python
# melt wide format -> long format
# seasons (autumn/winter/spring/summer) become a single 'season' column
# grades above 100 are retained as they may reflect extra credit or curved grading
grade_cols = ['autumn', 'winter', 'spring', 'summer']

df_long = df.melt(
    id_vars=['student_id', 'subject', 'year'],
    value_vars=grade_cols,
    var_name='season',
    value_name='grade'
)
```
