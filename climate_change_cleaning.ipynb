# Climate Change Data — Data Cleaning

## 0. Import & Load Data

```python
import pandas as pd
df = pd.read_excel('climate_change_edit.xls', engine='xlrd')
```

## 2. Data Cleaning

```python
# standardize column names
df.columns = df.columns.astype(str).str.strip().str.replace(' ', '_')
print(df.columns.tolist())
```

```python
# replace '..' with NaN
df = df.replace('..', pd.NA)
```

```python
# drop metadata columns
df = df.drop(columns=['SCALE', 'Decimals'])
```

```python
# convert year columns from object to numeric
year_cols = [str(y) for y in range(1990, 2012)]
df[year_cols] = df[year_cols].apply(pd.to_numeric)
df.dtypes
```

```python
# melt wide format -> long format
df_long = df.melt(
    id_vars=['Country_code', 'Country_name', 'Series_code', 'Series_name'],
    var_name='Year',
    value_name='Value'
)

df_long['Year'] = df_long['Year'].astype(int)
```
